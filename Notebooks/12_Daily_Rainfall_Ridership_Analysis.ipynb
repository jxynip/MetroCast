# Stage 12 — Daily Rainfall and Delhi Metro Ridership

This notebook documents the completed Stage 12 descriptive analysis. It only reads the saved Stage 12 CSV outputs and displays the saved figures; it does **not** rerun aggregation, statistical tests, or modelling.

## Methodology

- **Period:** 1 September 2024 through 28 February 2025 (181 calendar days).
- **Ridership outcome:** citywide daily total of all available hourly **Entry + Exit** counts.
- **Rainfall source:** existing Rajiv Chowk-aligned hourly `rain_mm` integration, summed by date.
- **Rainy-day definition:** daily rainfall > 0 mm over the available hours; otherwise dry.
- **Coverage:** every date has 21 hourly weather observations and 42 Entry/Exit ridership rows. The three unrepresented overnight hours are not included in either daily rainfall or ridership totals.
- **Tests reported:** Welch two-sample t-test, Mann–Whitney test, Cohen's *d*, weekday-matched descriptive comparison, weekday-adjusted OLS, and Pearson/Spearman rainfall associations.

The analysis is observational and descriptive. It does not identify a causal effect of rainfall.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Image

root = Path.cwd()
if not (root / 'DataProcessed').exists():
    root = root.parent
data = root / 'DataProcessed'
daily = pd.read_csv(data / 'stage12_daily_rainfall_ridership.csv', parse_dates=['date'])
results = pd.read_csv(data / 'stage12_daily_rainfall_ridership_results.csv')
display(results)
```

## Rainy versus dry days

There were **39 rainy** and **142 dry** days. Mean daily total Entry + Exit ridership was **6,121,841** on rainy days and **5,916,127** on dry days. The descriptive difference was **+205,714 riders** (**+3.48%** relative to dry days), with Cohen's *d* = **0.231**.

The overall difference was not statistically significant: Welch *t* = 1.383, *p* = 0.171; Mann–Whitney *p* = 0.1649.

```python
Image(filename=str(data / 'stage12_daily_rainfall_group_comparison.png'))
```

## Weekday-matched and adjusted results

All seven weekdays had both rainy and dry observations. Averaging the within-weekday rainy-minus-dry differences equally across weekdays produced **+174,336 riders** (**+2.95%**).

An OLS comparison controlling for weekday estimated a rainy-day difference of **+195,925 riders**, with *p* = **0.06231**. This is not statistically significant at the conventional 0.05 threshold. The adjusted comparison controls only for weekday; it does not account for holidays, service changes, seasonal trends, or other confounders.

```python
Image(filename=str(data / 'stage12_daily_rainfall_association.png'))
```

## Rainfall–ridership association

The association between daily rainfall amount and daily ridership was weak and not statistically significant:

- Pearson *r* = **0.114**, *p* = **0.1281**.
- Spearman rho = **0.119**, *p* = **0.1097**.

These correlations describe co-movement in this sample; they do not estimate a weather effect.

## Interpretation and limitations

The sample shows slightly higher mean recorded ridership on rainy days, but neither the overall comparison nor the weekday-adjusted comparison meets the conventional 0.05 significance threshold. The evidence therefore does not support a reliable daily-rainfall difference in this period.

Key limitations are the single-location observed-weather measure, incomplete overnight coverage, the relatively small rainy-day sample, and potential confounding from holidays, seasonality, special events, service variation, and rainfall timing/intensity. No causal conclusion should be drawn.
